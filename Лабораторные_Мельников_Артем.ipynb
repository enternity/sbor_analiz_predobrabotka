## «Numpy 2. Работа аэропорта NNN»

**Выполнил:** Мельников Артем Иванович, группа ЕТ-216

Цель: получить практические навыки работы с двумерным массивом NumPy, агрегирования, срезов, условной логики и булевой фильтрации, а также освоить операции с матрицами из модуля `linalg`.


````python
import numpy as np

np.random.seed(42)
n = 30

departures = np.random.randint(15, 65, size=n)
arrivals = np.random.randint(15, 65, size=n)

airport = np.column_stack(
    (departures, arrivals, departures + arrivals)
)

print("ВЫЛЕТ ПРИЛЁТ ВСЕГО")
print(airport)
print("Минимумы:", airport.min(axis=0))
print("Максимумы:", airport.max(axis=0))
````

````python
for i, name in enumerate(["ВЫЛЕТ", "ПРИЛЁТ", "ВСЕГО"]):
    print(name)
    print("Среднее:", airport[:, i].mean())
    print("Медиана:", np.median(airport[:, i]))
    print("Дисперсия:", airport[:, i].var())
    print("СКО:", airport[:, i].std())

half1 = airport[:15, 2].sum()
half2 = airport[15:, 2].sum()

if half1 > half2:
    print("Больше рейсов: первая половина")
elif half2 > half1:
    print("Больше рейсов: вторая половина")
else:
    print("Половины равны")

saturday = airport[5::7, 1]
print("Субботы:", saturday)
print("Вылеты > 10:", airport[airport[:, 0] > 10, 0])
print("10 < Вылеты < 50:",
      airport[(airport[:, 0] > 10) & (airport[:, 0] < 50), 0])
````

По данным отчёта:

- среднее число вылетов — **39,10**;
- среднее число прилётов — **38,43**;
- среднее общее число рейсов — **77,53**;
- дни 1–15: **1159** рейсов;
- дни 16–30: **1167** рейсов;
- субботы по прилётам: **[61, 53, 40, 22]**;
- вылетов больше 10 — **30**;
- вылетов больше 10, но меньше 50 — **21**.


````python
np.random.seed(42)

A = np.random.randint(1, 11, size=(3, 3))

even_elements = A[A % 2 == 0]
even_sum = even_elements.sum()

AAT = A @ A.T
AAT_inv = np.linalg.inv(AAT)
B = AAT @ AAT_inv

M = np.empty_like(A, dtype=float)
for i in range(3):
    for j in range(3):
        minor = np.delete(np.delete(A, i, axis=0), j, axis=1)
        M[i, j] = np.linalg.det(minor)

print("A:
", A)
print("Чётные элементы:", even_elements)
print("Сумма чётных:", even_sum)
print("A @ A.T:
", AAT)
print("(A @ A.T)^(-1):
", np.round(AAT_inv, 4))
print("B:
", np.round(B, 4))
print("det(B):", np.linalg.det(B))
print("M:
", M)
````

По отчёту чётные элементы матрицы **[4, 8, 10, 8]**, их сумма — **30**. Матрица `B = (A·Aᵀ)·(A·Aᵀ)⁻¹` численно совпадает с единичной с погрешностью порядка `10⁻¹⁴`, `det(B) ≈ 1`.
